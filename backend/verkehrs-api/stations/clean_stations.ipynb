# Data Cleaning for Vienna Stations

This notebook cleans the `vienna_stations_all_raw_rbl.csv` file.

```python
import pandas as pd
import os

# File path
input_file = "vienna_stations_all_raw_rbl.csv"

# Load data with correct delimiter
try:
    df = pd.read_csv(input_file, sep=';', encoding='windows-1252', on_bad_lines='skip')
    print("Data loaded successfully!")
    print(df.head())
except Exception as e:
    print(f"Error loading data: {e}")
```

```python
# Check for missing values
df.info()
```

```python
# Drop rows where essential information is missing if necessary, or just view them
# For example, if we need 'StopID' and 'StopText', we check for NaNs there.
missing_data = df[df.isnull().any(axis=1)]
print(f"Rows with missing data: {len(missing_data)}")
missing_data.head()
```

```python
# Filter for stations with a standalone 'U' in StopText (indicating U-Bahn station)
# The regex pattern r'\bU\b' looks for 'U' surrounded by word boundaries.
df_filtered = df[df['StopText'].str.contains(r'\bU\b', regex=True, na=False)]

print(f"Original number of stations: {len(df)}")
print(f"Filtered number of stations (with standalone 'U'): {len(df_filtered)}")
print(df_filtered[['StopText']].head())

output_file = "vienna_stations_filtered_u.csv"
df_filtered.to_csv(output_file, index=False)
print(f"Filtered data saved to {output_file}")
```
